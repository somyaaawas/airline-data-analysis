```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("../data/flights.csv", low_memory=False)

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
airline_counts = df["AIRLINE"].value_counts()

plt.figure(figsize=(10,6))
airline_counts.plot(kind="bar")

plt.title("Number of Flights per Airline")
plt.xlabel("Airline")
plt.ylabel("Flights")

plt.show()
```

```python
delay_airline = df.groupby("AIRLINE")["ARRIVAL_DELAY"].mean()

plt.figure(figsize=(10,6))
delay_airline.sort_values().plot(kind="bar")

plt.title("Average Arrival Delay by Airline")
plt.ylabel("Delay (minutes)")

plt.show()
```

```python
monthly_delay = df.groupby("MONTH")["ARRIVAL_DELAY"].mean()

plt.figure(figsize=(10,6))
sns.lineplot(x=monthly_delay.index, y=monthly_delay.values)

plt.title("Average Delay by Month")
plt.xlabel("Month")
plt.ylabel("Delay")

plt.show()
```

```python
busiest = df["ORIGIN_AIRPORT"].value_counts().head(10)

plt.figure(figsize=(10,6))
busiest.plot(kind="bar")

plt.title("Top 10 Busiest Airports")
plt.xlabel("Airport")
plt.ylabel("Flights")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df["ARRIVAL_DELAY"], bins=50)

plt.title("Distribution of Arrival Delays")

plt.show()
```

```python
numeric = df.select_dtypes(include=np.number)

plt.figure(figsize=(10,8))

sns.heatmap(numeric.corr(), cmap="coolwarm")

plt.title("Feature Correlation Heatmap")

plt.show()
```
